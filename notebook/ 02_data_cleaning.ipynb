```python
ids_mapping = pd.read_csv("../data/raw/IDS_mapping.csv")
```

```python
ids_mapping.columns
```

```python
ids_mapping.head(20)
```

```python
ids_mapping[
    ids_mapping["admission_type_id"].isin(
        ["discharge_disposition_id", "admission_source_id"]
    )
]
```

```python
discharge_start = ids_mapping[
    ids_mapping["admission_type_id"] == "discharge_disposition_id"
].index[0]

source_start = ids_mapping[
    ids_mapping["admission_type_id"] == "admission_source_id"
].index[0]

admission_map = ids_mapping.iloc[:discharge_start].copy()
discharge_map = ids_mapping.iloc[discharge_start + 1:source_start].copy()
source_map = ids_mapping.iloc[source_start + 1:].copy()
```

```python
admission_map.columns = ["code", "description"]
discharge_map.columns = ["code", "description"]
source_map.columns = ["code", "description"]
```

```python
admission_map = admission_map.dropna(subset=["code"])
discharge_map = discharge_map.dropna(subset=["code"])
source_map = source_map.dropna(subset=["code"])
```

```python
print("Admission types:")
display(admission_map)

print("\nDischarge dispositions:")
display(discharge_map)

print("\nAdmission sources:")
display(source_map)
```

```python
admission_map["code"] = pd.to_numeric(admission_map["code"])
discharge_map["code"] = pd.to_numeric(discharge_map["code"])
source_map["code"] = pd.to_numeric(source_map["code"])
```

```python
admission_dict = dict(
    zip(admission_map["code"], admission_map["description"])
)

discharge_dict = dict(
    zip(discharge_map["code"], discharge_map["description"])
)

source_dict = dict(
    zip(source_map["code"], source_map["description"])
)
```

```python
clean_df["admission_type"] = clean_df["admission_type_id"].map(admission_dict)

clean_df["discharge_disposition"] = clean_df["discharge_disposition_id"].map(discharge_dict)

clean_df["admission_source"] = clean_df["admission_source_id"].map(source_dict)
```
